# Badminton tekst notebook
In dit notebook laat ik zien hoe we bestanden kunnen inlezen en de data daarin kunnen manipuleren.

Voor dit notebook heb je de `badminton.txt` file nodig.

```python
f = open('badminton.txt', 'r')

content = f.read()

print(content)
```

```python
content = content.replace(".", "")
content = content.replace(",", "")
content = content.replace(")", "")
content = content.replace("(", "")
content = content.replace(";", "")
content = content.replace("\n", " ")
print(content)
```

```python
content = content.split(" ")
print(content)
```
